```python
import papermill as pm
import os
import pandas as pd
import scrapbook as sb
import sys
from pathlib import Path
import geopandas as gpd
```

```python
## notebook path
notebook_path = os.path.abspath(os.path.join('..', 'Notebooks', 'storm_tracking_results.ipynb'))
repo_root = Path.cwd() if (Path.cwd() / "functions").exists() else Path.cwd().parent
# add repo root so `import functions` works
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))
output_path = "storm_tracking_results_output.ipynb"
control_area_path = '/Users/diegoosorio/Library/CloudStorage/OneDrive-OklahomaAandMSystem/Academia/Codes/RainyDay_Runs/Run_CONUS/Shapes/'
transposition_domain_path = "/Users/diegoosorio/Library/CloudStorage/OneDrive-OklahomaAandMSystem/Academia/Codes/RainyDay_Runs/Run_CONUS/Domains/"
shp_path = "/Users/diegoosorio/Library/CloudStorage/OneDrive-OklahomaAandMSystem/Academia/Projects/Cartography/CONUS_5d_grid.shp"
grid = gpd.read_file(shp_path)
## parameters for the notebook
morph_radius = 2 # radius for morph structure
high_treshold = 0.5 # threshold for high precipitation (in mm/hr)
storm_interval = '12H' # storm time interval for getting the most intese precipitation

```

```python
## list of domains names
folder  = '/Users/diegoosorio/Library/CloudStorage/OneDrive-OklahomaAandMSystem/Academia/Codes/RainyDay_Runs/Run_CONUS/Domains'
domain_list = os.listdir(folder)
domain_list = [d for d in domain_list if d.endswith('.shp')]
domain_list = [d for d in domain_list if d != '.shp']

```

```python
conus_results = pd.DataFrame(columns=['angle', 'speed', 'intensity'])

for i, domain in grid.loc[:28].iterrows():
    # Extract region name from the domain file name
    domain_path = transposition_domain_path + "Domain_" + str(domain['id'])[:-1]+ "shp"
    
    print(i,domain['D_name'])
    pm.execute_notebook(
            notebook_path,        # Input notebook path
            'executed_notebook_temp.ipynb',       # Output notebook path
            parameters={
                'morph_radius': morph_radius,
                'high_treshold': high_treshold,
                'storm_interval': storm_interval,
                'control_area_path': control_area_path + "control_area_circle_" + "Domain_" + str(domain['id']) + '.shp',
                'transposition_domain_path': domain_path,
                'catalog_folder': f'/Users/diegoosorio/Library/CloudStorage/OneDrive-OklahomaAandMSystem/Academia/Codes/RainyDay_Runs/Run_CONUS/Results/Domain_{domain["id"]}_circle_10000km2/StormCatalog',
                'domain_name': domain['D_name']
                
            }
        )
    # Read the output notebook
    nb = sb.read_notebook('executed_notebook_temp.ipynb').scraps
    conus_results.loc[domain['D_name'],'angle'] = [nb['mean_angle'].data]
    conus_results.loc[domain['D_name'],'speed'] = [nb['mean_vel'].data]
    conus_results.loc[domain['D_name'],'intensity'] = [nb['mean_p'].data]

    # Clean up the temporary notebook
    os.remove('executed_notebook_temp.ipynb')
    # save results to a CSV file
```
